**Algorithm:** Advantage-Weighted Actor-Critic (AWAC)  
**Libraries:** `d3rlpy`, `minari`, `gymnasium`

### Download and Load Dataset:

```python
import d3rlpy
import minari

# Download dataset from Minari repository
minari.download_dataset("mujoco/invertedpendulum/medium-v0")

# Load offline dataset and environment into d3rlpy format
dataset, env = d3rlpy.datasets.get_minari("mujoco/invertedpendulum/medium-v0")
```

### Configure AWAC Algorithm:

```python
import torch

# Select the GPU if available, otherwise default to CPU
device_name = "cuda:0" if torch.cuda.is_available() else "cpu"

# Print the active processing hardware
print(f"Using device: {device_name}")

# Set up hyperparameter configuration for AWAC
config = d3rlpy.algos.AWACConfig(
    batch_size=256,
    actor_learning_rate=3e-4,
    critic_learning_rate=3e-4,
)

# Instantiate the model algorithm
awac = config.create(device=device_name)
```

### Model training on the offline dataset:

```python
awac.fit(
    dataset,
    n_steps=10000,
    n_steps_per_epoch=1000,
)
```

### Saving the model file:

```python
awac.save("awac_inverted_pendulum.d3")
```
